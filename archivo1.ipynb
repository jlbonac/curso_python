```python
import random

def jugar_piedra_papel_tijera():
    """
    Juega una partida de Piedra, Papel o Tijera contra la computadora.
    Solicita el nombre del jugador y utiliza condicionales para determinar el ganador.
    """
    nombre_jugador = input("¡Hola! ¿Cuál es tu nombre? ")
    print(f"¡Hola, {nombre_jugador}! Vamos a jugar Piedra, Papel o Tijera.")

    opciones = ["piedra", "papel", "tijera"]
    eleccion_computadora = random.choice(opciones)

    while True:
        eleccion_jugador = input("Elige piedra, papel o tijera: ").lower()
        if eleccion_jugador in opciones:
            break
        else:
            print("Opción inválida. Por favor, elige piedra, papel o tijera.")

    print(f"{nombre_jugador} eligió: {eleccion_jugador}")
    print(f"La computadora eligió: {eleccion_computadora}")

    if eleccion_jugador == eleccion_computadora:
        print("¡Empate!")
    elif (eleccion_jugador == "piedra" and eleccion_computadora == "tijera") or \
         (eleccion_jugador == "papel" and eleccion_computadora == "piedra") or \
         (eleccion_jugador == "tijera" and eleccion_computadora == "papel"):
        print(f"¡Felicidades, {nombre_jugador}! ¡Ganaste!")
    else:
        print("¡La computadora gana!")

if __name__ == "__main__":
    jugar_piedra_papel_tijera()
```

```python
import random

def obtener_jugada(nombre_jugador):
    """Solicita la jugada al jugador y la valida."""
    while True:
        jugada = input(f"{nombre_jugador}, elige (piedra, papel o tijera): ").lower()
        if jugada in ["piedra", "papel", "tijera"]:
            return jugada
        else:
            print("Jugada inválida. Por favor, elige entre piedra, papel o tijera.")

def determinar_ganador(jugada1, jugada2, nombre1, nombre2):
    """Determina el ganador de la ronda."""
    if jugada1 == jugada2:
        return "Empate"
    elif (jugada1 == "piedra" and jugada2 == "tijera") or \
         (jugada1 == "papel" and jugada2 == "piedra") or \
         (jugada1 == "tijera" and jugada2 == "papel"):
        return f"¡{nombre1} gana!"
    else:
        return f"¡{nombre2} gana!"

def jugar():
    """Función principal para ejecutar el juego."""
    nombre_jugador1 = input("Ingrese el nombre del Jugador 1: ")
    nombre_jugador2 = input("Ingrese el nombre del Jugador 2: ")

    print("\n¡Comienza el juego!")

    while True:
        jugada_jugador1 = obtener_jugada(nombre_jugador1)
        jugada_jugador2 = obtener_jugada(nombre_jugador2)

        print(f"{nombre_jugador1} eligió: {jugada_jugador1}")
        print(f"{nombre_jugador2} eligió: {jugada_jugador2}")

        resultado = determinar_ganador(jugada_jugador1, jugada_jugador2, nombre_jugador1, nombre_jugador2)
        print(resultado)

        jugar_otra_vez = input("¿Quieren jugar otra ronda? (sí/no): ").lower()
        if jugar_otra_vez != "sí":
            print("\n¡Gracias por jugar!")
            break

if __name__ == "__main__":
    jugar()
```
